```python
import numpy as np
import pandas as pd
```

```python
data = {1:[25],2:[40],3:[60],4:[85],5:[100],6:[95],7:[35]}
data={'hours_of_study':[1,2,3,4,5,6,7],'marks':[25,40,60,85,100,95,75]}
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(x=df['hours_of_study'],y=df['marks'])
```

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
pf=PolynomialFeatures(2)
```

```python
x = pf.fit_transform(df[['hours_of_study']])
```

```python
# 1^0=1,2^0=1,3,4,...
# 1^1=1,2^1=2,3,4,...
# 1^2=1,2^2=4,3,4,...
```

```python
y=df[['marks']]
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(x,y)
```

```python
model.predict([[1,5,25]])
```
