```python
for i in range(len(weight)):
    if(weight[i].find('kg')==-1):
        weight[i] = weight[i].replace(" lb", "")
        weight[i]=float(weight[i])
        weight[i] = weight[i]*0.453592
    elif(weight[i].find('lb')==-1):
        weight[i]=weight[i].replace(" kg", "")
        weight[i]=float(weight[i])
        
```
