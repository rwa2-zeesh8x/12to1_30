```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Connected_People": [
        "CUST-101",
        "CUST-102",
        "CUST-103",
        "EMP-101",
        "EMP-102",
        "CUST-104",
        "CUST-105"
    ],

    "Age": [
        25,
        31,
        35,
        38,
        np.nan,
        28,
        35
    ],

    "Weight": [
        "65 kg",
        "72 kg",
        "58 kg",
        "80 kg",
        np.nan,
        "75 kg",
        "69 kg"
    ],

    "Registration_Date": [
        "2024-05-10",
        "2025-08-21",
        "2025-12-15",
        "2026-01-20",
        "2026-03-05",
        "2025-01-01",
        "2023-02-01"
    ],

    "Salary": [
        35000,
        52000,
        41000,
        np.nan,
        65000,
        32000,
        45000
    ]
}

```

```python
is_emp = []
for i in range(len(df['Connected_People'])):
    if(df['Connected_People'][i].find("CUST") == -1):
        is_emp.append(1)
    else:
        is_emp.append(0)
```

```python
is_emp
```

```python
df['Is_Employee'] = is_emp
```

```python
df
```

```python
clean_rec = []
for i in range(len(df['Connected_People'])):
    if(df['Connected_People'][i].find("CUST") == -1):
        clean_rec.append(df['Connected_People'][i].replace("EMP-",""))
    elif(df['Connected_People'][i].find("EMP") == -1):
        clean_rec.append(df['Connected_People'][i].replace("CUST-",""))
```

```python
clean_rec
```

```python
df['id']=clean_rec
```

```python
df
```

```python
df.drop(columns='Connected_People',inplace=True)
```

```python
df
```

```python
df['Weight']=df['Weight'].fillna(df['Weight'].mode()[0])
```

```python
clean_weight = []

for i in range(len(df['Weight'])):
    clean_weight.append(df['Weight'][i].replace(" kg",""))
    
```
